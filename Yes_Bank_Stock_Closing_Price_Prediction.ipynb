### Hypothetical Statement - 2

- **Null Hypothisis** : The stock price gets effected by social news
- **Alternate Hypothisis**: The stock price does not gets effected by social news

Answer Here.

#### 2. Perform an appropriate statistical test.

```python
# Perform Statistical Test to obtain P-Value
data_before_crash = data.loc[:'2018-01-01']
data_after_crash = data.loc['2018-01-01':]
```

```python
from scipy import stats

# Extract 'Close' prices for both periods
close_prices_before = data_before_crash['Close']
close_prices_after = data_after_crash['Close']

# Perform independent two-sample t-test
t_statistic, p_value = stats.ttest_ind(close_prices_before, close_prices_after, equal_var=False) # Assuming unequal variances

print(f"T-statistic: {t_statistic:.4f}")
print(f"P-value: {p_value:.4e}")
```

##### Which statistical test have you done to obtain P-Value?

An independent two-sample t-test was performed.

Given the extremely small p-value (1.3973e-25), which is significantly less than the common significance level of 0.05, we reject the null hypothesis. This indicates that there is a statistically significant difference in the mean closing prices of Yes Bank's stock before and after the period associated with the social news (Rana Kapoor fraud case).

**Conclusion:** The stock price is significantly affected by social news events.

##### Why did you choose the specific statistical test?

I chose the independent two-sample t-test because we are comparing the means of two independent groups (stock prices before the event and stock prices after the event) to determine if there's a statistically significant difference between them. This test is suitable for assessing whether the 'social news' event (the fraud case) had an effect on the stock prices, as indicated by a change in the average closing price.

### Hypothetical Statement - 3

**Null Hypothisis :** The mean Normalized Closing Position (NCP)
during the growth/bullish era (July 2005 to December 2017) is less than or equal to the mean Normalized Closing Position post-2018

**Alternate Hypothisis :** The mean Normalized Closing Position (NCP)
during the growth/bullish era (July 2005 to December 2017) is significantly higher than the mean Normalized Closing Position post-2018

Answer Here.

#### 2. Perform an appropriate statistical test.

```python
import numpy as np
import pandas as pd
from scipy import stats

data_hyp3 = data.sort_values("Dates")

data_hyp3["NCP"] = (data_hyp3["Close"] - data_hyp3["Low"]) / (data_hyp3["High"] - data_hyp3["Low"])
data_hyp3
```

```python
# 3. Split into two historical periods
# from datetime import date
pre_2018 = data_hyp3[data_hyp3.index < datetime.fromisoformat("2018-01-01")]["NCP"]
post_2018 = data_hyp3[data_hyp3.index >= datetime.fromisoformat("2018-01-01")]["NCP"]

# 4. Perform a One-Tailed Welch's t-test
t_stat, p_val = stats.ttest_ind(pre_2018, post_2018, equal_var=False)

# One-tailed p-value calculation
p_val_one_tailed = p_val / 2 if t_stat > 0 else 1 - (p_val / 2)

print(f"Pre-2018 Mean NCP: {pre_2018.mean():.4f}")
print(f"Post-2018 Mean NCP: {post_2018.mean():.4f}")
print(f"T-statistic: {t_stat:.4f}")
print(f"P-value (one-tailed): {p_val_one_tailed:.4e}")
```

##### Which statistical test have you done to obtain P-Value?

One-Tailed Test

##### Why did you choose the specific statistical test?

Here there are two samples. However two sample test is used then we only want to know that two  sample means are differents irrespective of the fact that which one is larger. Here, one tailed test is used because financial domain knowledge gives a clear directional prediction that growth era NCP will be more than Cricis era NCP before running the data.

1st Differencing

```python
data_diff = data['Close'].diff().dropna()
```

```python
from statsmodels.tsa.stattools import adfuller

# Perform the Augmented Dickey-Fuller test on the differenced data
result_diff = adfuller(data_diff)

# Print the results
print('ADF Statistic (Differenced Data): %f' % result_diff[0])
print('p-value (Differenced Data): %f' % result_diff[1])
print('Critical Values:')
for key, value in result_diff[4].items():
    print('\t%s: %.3f' % (key, value))

# Interpret the results
if result_diff[1] <= 0.05:
    print('Conclusion (Differenced Data): The differenced time series is likely stationary (reject H0).')
else:
    print('Conclusion (Differenced Data): The differenced time series is likely non-stationary (fail to reject H0).')
```

second differencing

```python
from statsmodels.tsa.stattools import adfuller

# Perform 2nd differencing (first differencing of the first differenced data)
data_diff2 = data['Close'].diff().diff().dropna()

# Perform the Augmented Dickey-Fuller test on the second differenced data
result_diff2 = adfuller(data_diff2)

# Print the results
print('ADF Statistic (2nd Differenced Data): %f' % result_diff2[0])
print('p-value (2nd Differenced Data): %f' % result_diff2[1])
print('Critical Values:')
for key, value in result_diff2[4].items():
    print('\t%s: %.3f' % (key, value))

# Interpret the results
if result_diff2[1] <= 0.05:
    print('Conclusion (2nd Differenced Data): The second differenced time series is likely stationary (reject H0).')
else:
    print('Conclusion (2nd Differenced Data): The second differenced time series is likely non-stationary (fail to reject H0).')
```

### ML Model - 1

#### 1. Explain the ML Model used and it's performance using Evaluation metric Score Chart.

```python
import warnings
from statsmodels.tools.sm_exceptions import ValueWarning

# Suppress only the specific ValueWarning from statsmodels
warnings.simplefilter('ignore', ValueWarning)
# ML Model - 1 Implementation: ARIMA
from statsmodels.tsa.arima.model import ARIMA
from sklearn.metrics import mean_squared_error, mean_absolute_error
import numpy as np

arima_order = (1, 2, 1)

# Fit the ARIMA model to the training data
# We will use the 'Close' prices for modeling
model_arima = ARIMA(train_data['Close'], order=arima_order)
model_arima_fit = model_arima.fit()

# Predict on the model for the test period
# The start and end indices for prediction correspond to the test_data
start_index = len(train_data)
end_index = len(data) - 1

predictions_arima = model_arima_fit.predict(start=start_index, end=end_index, typ='levels')

# Create a DataFrame for predictions to align with test_data index
predictions_arima.index = test_data.index

# Evaluate the model
mse_arima = mean_squared_error(test_data['Close'], predictions_arima)
mae_arima = mean_absolute_error(test_data['Close'], predictions_arima)
rmse_arima = np.sqrt(mse_arima)

print(f"\nARIMA({arima_order}) Mean Absolute Error (MSE): {mae_arima:.4f}")
print(f"\nARIMA({arima_order}) Mean Squared Error (MSE): {mse_arima:.4f}")
print(f"ARIMA({arima_order}) Root Mean Squared Error (RMSE): {rmse_arima:.4f}")

# You can visualize the predictions against actual values if desired
plt.figure(figsize=(14, 5))
plt.plot(train_data['Close'], label='Training Data')
plt.plot(test_data['Close'], label='Actual Close Prices', color='orange')
plt.plot(predictions_arima, label='ARIMA Predictions', color='green', linestyle='--')
plt.title(f'ARIMA({arima_order}) Model - Actual vs. Predicted Close Prices')
plt.xlabel('Date')
plt.ylabel('Close Price')
plt.legend()
plt.grid(True)
plt.show()
```

#### 2. Cross- Validation & Hyperparameter Tuning

```python
import warnings
from statsmodels.tools.sm_exceptions import ValueWarning

# Suppress only the specific ValueWarning from statsmodels
warnings.simplefilter('ignore', ValueWarning)
# ML Model - 1 Implementation with hyperparameter optimization techniques (GridSearch)
from statsmodels.tsa.arima.model import ARIMA
from sklearn.metrics import mean_squared_error, mean_absolute_error
import numpy as np
import itertools

# Define the range for p and q parameters (d is fixed at 2)
p_values = range(0, 3)  # Example range for p
q_values = range(0, 3)  # Example range for q
d_value = 2  # Fixed as determined from ADF test

# Generate all possible combinations of p, d, q
orders = list(itertools.product(p_values, [d_value], q_values))

best_mae_arima = float('inf')
best_order_arima = None

print('Searching for optimal ARIMA parameters...')

for order in orders:
    try:
        # Fit the ARIMA model
        model_arima = ARIMA(train_data['Close'], order=order)
        model_arima_fit = model_arima.fit()

        # Make predictions on the test set
        start_index = len(train_data)
        end_index = len(data) - 1
        predictions = model_arima_fit.predict(start=start_index, end=end_index, typ='levels')
        predictions.index = test_data.index

        # Evaluate the model
        mae = mean_absolute_error(test_data['Close'], predictions)

        # Check if current model is better
        if mae < best_mae_arima:
            best_mae_arima = mae
            best_order_arima = order

        print(f'ARIMA{order} RMSE: {mae:.4f}')
    except Exception as e:
        # Catch potential errors during model fitting (e.g., non-invertible, non-stationary)
        print(f'ARIMA{order} failed: {e}')
        continue

print(f'\nBest ARIMA Order: {best_order_arima} with RMSE: {best_mae_arima:.4f}')

# After finding the best order, fit the final model and make predictions
if best_order_arima:
    final_model_arima = ARIMA(train_data['Close'], order=best_order_arima)
    final_model_arima_fit = final_model_arima.fit()

    # Predict on the model for the test period
    start_index = len(train_data)
    end_index = len(data) - 1
    predictions_arima_tuned = final_model_arima_fit.predict(start=start_index, end=end_index, typ='levels')
    predictions_arima_tuned.index = test_data.index

    # Evaluate the final tuned model
    mae_arima_tuned = mean_absolute_error(test_data['Close'], predictions_arima_tuned)
    # rmse_arima_tuned = np.sqrt(mse_arima_tuned)

    print(f"\nARIMA (tuned {best_order_arima}) Mean Absolute Error (MAE): {mae_arima_tuned:.4f}")

    # Plotting the tuned ARIMA predictions
    plt.figure(figsize=(14, 5))
    plt.plot(train_data['Close'], label='Training Data')
    plt.plot(test_data['Close'], label='Actual Close Prices', color='orange')
    plt.plot(predictions_arima_tuned, label=f'ARIMA Tuned Predictions {best_order_arima}', color='green', linestyle='--')
    plt.title(f'ARIMA Tuned Model - Actual vs. Predicted Close Prices (Order: {best_order_arima})')
    plt.xlabel('Date')
    plt.ylabel('Close Price')
    plt.legend()
    plt.grid(True)
    plt.show()
```

```python
# ML Model - 3 Implementation with hyperparameter optimization techniques (GridSearch) for XGBoost
import itertools
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from xgboost import XGBRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error
from sklearn.model_selection import GridSearchCV, TimeSeriesSplit

# 1. Feature Engineering (Using features engineered in Cell 1)
df_xgb = data.copy()
df_xgb['Close_Lag1'] = df_xgb['Close'].shift(1)
df_xgb['Close_Lag2'] = df_xgb['Close'].shift(2)
df_xgb['MA_3'] = df_xgb['Close'].rolling(window=3).mean()
df_xgb['MA_6'] = df_xgb['Close'].rolling(window=6).mean()
df_xgb['High_Low_Spread'] = df_xgb['High'] - df_xgb['Low']
df_xgb['Open_Close_Diff'] = df_xgb['Open'] - df_xgb['Close']
df_xgb = df_xgb.dropna()

# Define Features and Target
features = ['Open', 'High', 'Low', 'Close_Lag1', 'Close_Lag2', 'MA_3', 'MA_6', 'High_Low_Spread', 'Open_Close_Diff']
X = df_xgb[features]
y = df_xgb['Close']

# Train-Test Chronological Split
train_size = len(train_data) - (len(data) - len(df_xgb))
X_train, X_test = X.iloc[:train_size], X.iloc[train_size:]
y_train, y_test = y.iloc[:train_size], y.iloc[train_size:]

# 2. Define Hyperparameter Search Grid
param_grid = {
    'n_estimators': [50, 100, 150],
    'max_depth': [2, 3, 5],
    'learning_rate': [0.01, 0.05, 0.1],
    'subsample': [0.8, 1.0],
    'colsample_bytree': [0.8, 1.0]
}

# 3. Setup TimeSeriesSplit for Cross-Validation (Prevents future data leakage)
tscv = TimeSeriesSplit(n_splits=3)

xgb_base = XGBRegressor(random_state=42)

print('Searching for optimal XGBoost parameters...')
grid_search = GridSearchCV(
    estimator=xgb_base,
    param_grid=param_grid,
    cv=tscv,
    scoring='neg_mean_absolute_error',
    n_jobs=-1,
    verbose=0
)

# Fit Grid Search on Training Data
grid_search.fit(X_train, y_train)

best_params_xgb = grid_search.best_params_
print(f'\nBest XGBoost Parameters: {best_params_xgb}')

# 4. Train Final Model with Optimal Parameters
final_model_xgb = XGBRegressor(**best_params_xgb, random_state=42)
final_model_xgb.fit(X_train, y_train)

# Predict on Test Set
predictions_xgb_tuned = final_model_xgb.predict(X_test)
predictions_xgb_tuned = pd.Series(predictions_xgb_tuned, index=y_test.index)

# 5. Evaluate Tuned Model
mae_xgb_tuned = mean_absolute_error(y_test, predictions_xgb_tuned)
rmse_xgb_tuned = np.sqrt(mean_squared_error(y_test, predictions_xgb_tuned))

print(f"\nXGBoost (Tuned) Mean Absolute Error (MAE): {mae_xgb_tuned:.4f}")
print(f"XGBoost (Tuned) Root Mean Squared Error (RMSE): {rmse_xgb_tuned:.4f}")

# 6. Plotting Tuned XGBoost Predictions
plt.figure(figsize=(14, 5))
plt.plot(y_train, label='Training Data')
plt.plot(y_test, label='Actual Close Prices', color='orange')
plt.plot(predictions_xgb_tuned, label='XGBoost Tuned Predictions', color='red', linestyle='--')
plt.title('XGBoost Tuned Model - Actual vs. Predicted Close Prices')
plt.xlabel('Date')
plt.ylabel('Close Price')
plt.legend()
plt.grid(True)
plt.show()
```
